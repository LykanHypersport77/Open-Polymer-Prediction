```python
import pandas as pd
```

```python


def process_data(file_path):
    data = []
    with open(file_path, 'r') as file:
        row_data = []
        for line in file:
            # Check if the line contains a class label
            if line.strip().isalpha():
                if row_data:  # if row_data is not empty, append it to data
                    data.append(row_data)
                current_class = line.strip()
                row_data = [current_class]  # start a new row with the class label
            else:
                # Process numerical data lines
                values = line.split()
                if values:
                    try:
                        # Convert each value to float and extend the current row data
                        row_data.extend(map(float, values))
                    except ValueError:
                        # Handle the case where conversion to float fails
                        print(f"Non-numeric value encountered: {values}")
        if row_data:  # append the last row if it's not empty
            data.append(row_data)

    return pd.DataFrame(data)

# Process the data and convert to DataFrame
df = process_data('lp1.txt')
df.columns = ColumnNames

# Set the column names
num_values = df.shape[1] - 1  # number of values per class


print(df)
```
